```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
iris = sns.load_dataset('iris')
```

```python
#Created a pairplot of the data set
# Setosa is the most separable 
sns.pairplot(iris,hue='species',palette='Dark2')
plt.show()
```

```python
#Create a kde plot of sepal_length versus sepal width for setosa species of flower
setosa = iris[iris['species'] == 'setosa']
sns.kdeplot(x=setosa['sepal_width'],y=setosa['sepal_length'],cmap="plasma",fill=True)
plt.show()
```

```python
#Split your data into a training set and a testing set
from sklearn.model_selection import train_test_split
```

```python
X = iris.drop('species',axis=1)
y = iris['species']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30)
```

```python
#Call the SVC() model from sklearn and fit the model to the training data.
```

```python
from sklearn.svm import SVC
```

```python
svc_model = SVC()
```

```python
svc_model.fit(X_train,y_train)
```

```python
#predictions from the model and create a confusion matrix and a classification report
```

```python
predictions = svc_model.predict(X_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
```

```python
print(confusion_matrix(y_test,predictions))
```

```python
print(classification_report(y_test,predictions))
```

```python
#Import GridsearchCV from SciKit Learn
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
#Created a dictionary called param_grid and fill out some parameters for C and gamma
param_grid = {'C': [0.1,1, 10, 100], 'gamma': [1,0.1,0.01,0.001]} 
```

```python
#Created a GridSearchCV object and fit it to the training data
grid = GridSearchCV(SVC(),param_grid,refit=True,verbose=2)
grid.fit(X_train,y_train)
```

```python
grid_predictions = grid.predict(X_test)
```

```python
print(confusion_matrix(y_test,grid_predictions))
```

```python
print(classification_report(y_test,grid_predictions))
```
